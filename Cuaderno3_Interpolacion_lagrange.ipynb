## ** Ejercicio 1: Bloque 09: Interpolación de Lagrange (2 Nodos / Grado 1)**

### **Contexto de Ingeniería Civil / Geotecnia (Asentamiento de Arcillas)**
Durante un ensayo de consolidación en laboratorio, se mide el asentamiento $s(t)$ (mm) en función del tiempo transcurrido $t$ (h) sobre un estrato de suelo arcilloso saturado:

| $i$ | Tiempo, $t_i$ (h) | Asentamiento, $s(t_i)$ (mm) |
| :---: | :---: | :---: |
| 0 | 1 | 0.000000 |
| 1 | 4 | 1.386294 |

**Se pide:**
* Estimar el asentamiento para un tiempo intermedio $t = 2\text{ h}$ mediante un polinomio de Lagrange de grado 1 ($n=2$ nodos).
* Calcular el valor exacto según el modelo empírico de referencia $s(t) = \ln(t)$ y determinar los errores absoluto ($E_a$) y relativo porcentual ($E_r\%$).

### **Resumen del Desarrollo y Comprobación Manual (Laura & Dana)**

1. **Bases de Lagrange Evaluadas ($t=2\text{ h}$):**
   $$L_0(2) = \frac{2 - 4}{1 - 4} = \frac{2}{3} \approx 0.666667, \quad L_1(2) = \frac{2 - 1}{4 - 1} = \frac{1}{3} \approx 0.333333$$
2. **Evaluación Polinomial:**
   $$P_1(2) = L_0(2) \cdot 0.000000 + L_1(2) \cdot 1.386294 = \frac{1}{3} (1.386294) = \mathbf{0.462098\text{ mm}}$$
3. **Control de Errores frente a Modelo Analítico $s(t) = \ln(t)$:**
   * $s_{\text{exacto}}(2) = \ln(2) \approx \mathbf{0.693147\text{ mm}}$.
   * Error Absoluto: $E_a = |0.693147 - 0.462098| = \mathbf{0.231049\text{ mm}}$.
   * Error Relativo Porcentual: $E_r\% = \frac{0.231049}{0.693147} \times 100\% = \mathbf{33.33\%}$.
4. **Interpretación Geotécnica:** La recta lineal subestima significativamente el asentamiento inicial debido a la marcada curvatura logarítmica del proceso de consolidación primaria en suelos arcillosos.

```python
# Datos de entrada del ensayo geotécnico
t_nodes_ex7 = np.array([1.0, 4.0])
s_nodes_ex7 = np.array([0.0, 1.386294])
t_consulta = 2.0

# Evaluación mediante el módulo modular
res_ex7 = evaluar_lagrange(t_nodes_ex7, s_nodes_ex7, t_consulta)

# Imprimir reporte completo
reportar_lagrange(res_ex7, t_nodes_ex7, s_nodes_ex7, t_consulta, unidad_x="h", unidad_y="mm")

# Cálculo de exactitud frente al modelo de referencia s(t) = ln(t)
val_exacto_ex7 = np.log(t_consulta)
error_abs_ex7 = abs(val_exacto_ex7 - res_ex7['y_eval'])
error_rel_pct_ex7 = (error_abs_ex7 / val_exacto_ex7) * 100.0

print("\n" + "=" * 75)
print("CUANTIFICACIÓN DE ERRORES (MODELO DE REFERENCIA s(t) = ln(t)):")
print(f"  Valor Exacto s(2 h)          = {val_exacto_ex7:.6f} mm")
print(f"  Estimación Lagrange P_1(2 h) = {res_ex7['y_eval']:.6f} mm")
print(f"  Error Absoluto (E_a)         = {error_abs_ex7:.6f} mm")
print(f"  Error Relativo Porcentual    = {error_rel_pct_ex7:.2f}%")
print("=" * 75)
```

## ** Ejercicio 2: Bloque 09: Interpolación de Lagrange (3 Nodos / Grado 2)**

### **Contexto de Ingeniería Civil / Estructuras (Prueba de Carga en Viga)**
Durante una prueba de carga en el centro de luz de una viga simplemente apoyada de concreto reforzado, se registraron las siguientes lecturas de carga $P$ (kN) vs. deflexión vertical $\delta$ (mm):

| $i$ | Carga, $P_i$ (kN) | Deflexión, $\delta(P_i)$ (mm) |
| :---: | :---: | :---: |
| 0 | 1 | 2 |
| 1 | 3 | 6 |
| 2 | 4 | 24 |

**Se pide:**
* Estimar la deflexión de la viga para una carga intermedia $P = 2\text{ kN}$ con un polinomio de Lagrange de segundo grado ($n=3$ nodos).
* Comprobar la reproducción exacta de nodos y realizar un análisis crítico sobre la validez física del resultado.

### **Resumen del Desarrollo y Comprobación Manual (Laura & Dana)**

1. **Bases de Lagrange Evaluadas ($P=2\text{ kN}$):**
   $$L_0(2) = \frac{(2-3)(2-4)}{(1-3)(1-4)} = \frac{1}{3}, \quad L_1(2) = \frac{(2-1)(2-4)}{(3-1)(3-4)} = 1, \quad L_2(2) = \frac{(2-1)(2-3)}{(4-1)(4-3)} = -\frac{1}{3}$$
2. **Evaluación Polinomial:**
   $$P_2(2) = \frac{1}{3}(2) + 1(6) - \frac{1}{3}(24) = \frac{2}{3} + 6 - 8 = -\frac{4}{3} \approx \mathbf{-1.333333\text{ mm}}$$
3. **Comprobación de Nodos:**
   * $P_2(1) = 2\text{ mm} \quad \checkmark$
   * $P_2(3) = 6\text{ mm} \quad \checkmark$
   * $P_2(4) = 24\text{ mm} \quad \checkmark$
4. **Análisis Crítico Estructural:**
   * **Incongruencia Física:** El signo negativo arrojado ($-1.333\text{ mm}$) implica que la viga se flexionaría hacia arriba bajo una carga descendente, violando el equilibrio estático.
   * **Causa Analítica:** Entre $3\text{ kN}$ y $4\text{ kN}$ la deflexión se dispara bruscamente ($6 \rightarrow 24\text{ mm}$) debido a la fisuración o plastificación del concreto. Al forzar un polinomio parabólico global a ajustarse a esta discontinuidad, la curva sufre un fenómeno de oscilación que destruye la precisión en el intervalo inicial.

```python
# Datos de la prueba de carga
P_nodes_ex8 = np.array([1.0, 3.0, 4.0])
d_nodes_ex8 = np.array([2.0, 6.0, 24.0])
P_consulta = 2.0

# Evaluación mediante el módulo modular
res_ex8 = evaluar_lagrange(P_nodes_ex8, d_nodes_ex8, P_consulta)

# Imprimir reporte completo
reportar_lagrange(res_ex8, P_nodes_ex8, d_nodes_ex8, P_consulta, unidad_x="kN", unidad_y="mm")
```

## ** Ejercicio 3: Bloque 09: Interpolación de Lagrange (4 Nodos / Comparación de Grados 3, 2 y 1)**

### **Contexto de Ingeniería Civil / Materiales y Estructuras (Evolución de $f'_c$)**
En una obra se ensayan cilindros de concreto curados a compresión $f'_c$ (psi) a distintas edades $t$ (días) para autorizar el desencofrado de una losa:

| $i$ | Edad, $t_i$ (días) | Resistencia $f'_c$ (psi) |
| :---: | :---: | :---: |
| 0 | 3 | 2310 |
| 1 | 5 | 3090 |
| 2 | 7 | 3940 |
| 3 | 13 | 4755 |

El ensayo de $t = 10\text{ días}$ falló. Se solicita estimar $f'_c(10\text{ días})$ construyendo y comparando polinomios de **Grado 3** (4 nodos), **Grado 2** (3 nodos: $t=5,7,13$) y **Grado 1** (2 nodos: $t=7,13$).

###  **Resumen del Desarrollo y Comprobación Manual (Laura & Dana)**

1. **Desarrollo Grado 3 (4 Nodos):**
   * Ponderaciones evaluadas: $L_0(10) = \frac{9}{16}, L_1(10) = -\frac{63}{32}, L_2(10) = \frac{35}{16}, L_3(10) = \frac{7}{32}$.
   * Valor interpolado: $P_3(10) = \mathbf{4874.8438\text{ psi}}$.
2. **Desarrollo Grado 2 (3 Nodos: $t=5,7,13$):**
   * Ponderaciones evaluadas: $L_1(10) = -\frac{9}{16}, L_2(10) = \frac{15}{12}, L_3(10) = \frac{15}{48}$.
   * Valor interpolado: $P_2(10) = \mathbf{4672.8125\text{ psi}}$.
3. **Desarrollo Grado 1 (2 Nodos: $t=7,13$):**
   * Ponderaciones evaluadas: $L_2(10) = \frac{1}{2}, L_3(10) = \frac{1}{2}$.
   * Valor interpolado: $P_1(10) = \mathbf{4347.5000\text{ psi}}$.
4. **Análisis Crítico y Admisibilidad Física:**
   * La resistencia a la compresión del concreto es monótonamente creciente. A los 10 días, $f'_c$ debe ser estrictamente superior a $f'_c(7\text{ días}) = 3940\text{ psi}$ e inferior a $f'_c(13\text{ días}) = 4755\text{ psi}$.
   * **Inadmisibilidad del Grado 3:** Supera erróneamente la resistencia del día 13 ($4874.84 > 4755\text{ psi}$) debido al sobreajuste u oscilaciones numéricas de polinomios de alto grado.
   * **Selección Técnica:** Los polinomios de Grado 1 ($4347.50\text{ psi}$) y Grado 2 ($4672.81\text{ psi}$) son físicamente admisibles, correspondiendo Grado 2 al modelo más preciso según la curva cóncava típica de hidratación del cemento.

```python
# Datos del proceso de curado
t_nodes_ex9 = np.array([3.0, 5.0, 7.0, 13.0])
fc_nodes_ex9 = np.array([2310.0, 3090.0, 3940.0, 4755.0])
t_target = 10.0

# 1. Grado 3 (4 nodos completos)
res_ex9_g3 = evaluar_lagrange(t_nodes_ex9, fc_nodes_ex9, t_target)

# 2. Grado 2 (3 nodos: t=5, 7, 13)
res_ex9_g2 = evaluar_lagrange(t_nodes_ex9[1:], fc_nodes_ex9[1:], t_target)

# 3. Grado 1 (2 nodos: t=7, 13)
res_ex9_g1 = evaluar_lagrange(t_nodes_ex9[2:], fc_nodes_ex9[2:], t_target)

print("=" * 75)
print("EVALUACIÓN DE INTERPOLACIÓN A t = 10 DÍAS (RESISTENCIA f'c)")
print("=" * 75)
print(f"Polinomio Grado 3 (4 Nodos: 3, 5, 7, 13 días) : P_3(10) = {res_ex9_g3['y_eval']:.4f} psi")
print(f"Polinomio Grado 2 (3 Nodos: 5, 7, 13 días)    : P_2(10) = {res_ex9_g2['y_eval']:.4f} psi")
print(f"Polinomio Grado 1 (2 Nodos: 7, 13 días)       : P_1(10) = {res_ex9_g1['y_eval']:.4f} psi")
print("=" * 75)

# Reporte detallado para Grado 3
reportar_lagrange(res_ex9_g3, t_nodes_ex9, fc_nodes_ex9, t_target, unidad_x="días", unidad_y="psi")
```
