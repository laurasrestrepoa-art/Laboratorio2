```python
import ipywidgets as widgets
from IPython.display import display, HTML, Markdown
import nbformat
import os

custom_ui_css = """
<style>
    @import url('https://googleapis.com');

    .voila-app { 
        background-color: #f8fafc !important; 
        font-family: 'Plus Jakarta Sans', -apple-system, sans-serif; 
        padding: 0 !important; 
    }
    
    .header-banner {
        background: linear-gradient(135deg, #1e3a8a 0%, #3b82f6 100%);
        color: white; 
        padding: 40px; 
        margin-bottom: 30px;
        box-shadow: 0 10px 25px rgba(30, 58, 138, 0.15);
        border-bottom-left-radius: 24px;
        border-bottom-right-radius: 24px;
    }
    .header-banner h1 { 
        color: white !important; 
        margin: 0 0 12px 0; 
        font-size: 32px; 
        font-weight: 700; 
        letter-spacing: -0.5px;
    }
    .header-banner p { 
        margin: 0; 
        font-size: 14.5px; 
        opacity: 0.9; 
        line-height: 1.6;
    }

    .widget-tab, .jupyter-widgets.widget-tab { 
        border: none !important; 
        background: transparent !important; 
        box-shadow: none !important;
    }
    .widget-tab > div:last-child, .lm-TabPanel-tabContents, .p-TabPanel-tabContents {
        border: none !important;
        background: transparent !important;
    }
    .widget-tab > .lm-TabBar, .widget-tab > .p-TabBar { 
        background: #ffffff !important; 
        padding: 10px 40px 0 40px !important; 
        border-bottom: 2px solid #e2e8f0 !important; 
    }
    .widget-tab .lm-TabBar-tab, .widget-tab .p-TabBar-tab {
        background: transparent !important; 
        color: #64748b !important; 
        font-weight: 600 !important; 
        padding: 14px 24px !important; 
        margin-right: 12px !important; 
        border: none !important; 
        font-size: 15px !important;
        transition: all 0.25s ease !important;
        position: relative !important;
        border-top-left-radius: 12px !important;
        border-top-right-radius: 12px !important;
    }
    .widget-tab .lm-TabBar-tab:hover, .widget-tab .p-TabBar-tab:hover {
        color: #2563eb !important;
        background: #f1f5f9 !important;
    }
    .widget-tab .lm-TabBar-tab.lm-mod-current, .widget-tab .p-TabBar-tab.p-mod-current { 
        background: #f8fafc !important; 
        color: #1e40af !important; 
        font-weight: 700 !important;
        border: none !important;
    }
    .widget-tab .lm-TabBar-tab.lm-mod-current::after, .widget-tab .p-TabBar-tab.p-mod-current::after {
        content: ''; 
        position: absolute; 
        bottom: 0; 
        left: 15%; 
        right: 15%; 
        height: 4px; 
        background-color: #2563eb !important;
        border-radius: 50px !important;
    }
    
    .card-panel {
        background-color: #ffffff; 
        border-radius: 20px;
        padding: 35px; 
        margin: 25px 40px;
        box-shadow: 0 4px 20px rgba(0, 0, 0, 0.03);
        border: 1px solid #e2e8f0;
        transition: transform 0.3s ease, box-shadow 0.3s ease;
    }
    .card-panel:hover {
        transform: translateY(-2px);
        box-shadow: 0 10px 30px rgba(0, 0, 0, 0.06);
    }
    
    .code-display { 
        background-color: #0f172a; 
        color: #e2e8f0; 
        padding: 20px; 
        border-radius: 16px; 
        font-family: 'JetBrains Mono', monospace; 
        font-size: 13.5px; 
        margin-bottom: 16px; 
        line-height: 1.6;
        box-shadow: inset 0 2px 8px rgba(0,0,0,0.2);
    }
    
    .console-output { 
        background-color: #f8fafc; 
        border-left: 5px solid #3b82f6; 
        padding: 20px; 
        margin-top: 15px; 
        border-radius: 12px; 
        font-family: 'JetBrains Mono', monospace; 
        font-size: 13.5px;
        color: #334155;
        box-shadow: 0 2px 8px rgba(0,0,0,0.02);
    }
    
    .btn-action { 
        background: linear-gradient(135deg, #2563eb 0%, #1d4ed8 100%) !important; 
        color: white !important; 
        border: none !important; 
        padding: 10px 24px !important; 
        height: auto !important; 
        min-height: 40px !important;
        line-height: normal !important;
        display: inline-flex !important; 
        align-items: center !important;
        justify-content: center !important;
        border-radius: 50px !important; 
        font-weight: 600 !important; 
        font-size: 14px !important;
        cursor: pointer; 
        transition: all 0.2s ease !important;
        box-shadow: 0 4px 12px rgba(37, 99, 235, 0.2);
    }
    .btn-action:hover { 
        transform: translateY(-1px);
        box-shadow: 0 6px 16px rgba(37, 99, 235, 0.35);
        background: linear-gradient(135deg, #1d4ed8 0%, #1e40af 100%) !important;
    }
    
    .btn-reset { 
        background-color: #f1f5f9 !important; 
        color: #475569 !important; 
        border: 1px solid #e2e8f0 !important; 
        padding: 10px 24px !important; 
        height: auto !important; 
        min-height: 40px !important;
        line-height: normal !important;
        display: inline-flex !important; 
        align-items: center !important;
        justify-content: center !important;
        border-radius: 50px !important; 
        font-weight: 600 !important;
        font-size: 14px !important;
        cursor: pointer; 
        transition: all 0.2s ease !important;
    }
    .btn-reset:hover {
        background-color: #e2e8f0 !important;
        color: #1e293b !important;
    }
</style>
"""
display(HTML(custom_ui_css))

banner_html = """
<div class="header-banner">
    <h1>Sistemas de Ecuaciones e Interpolación</h1>
    <p><b>Integrantes:</b> Laura Sofía Restrepo Ardila y Dana Sarai Soto Colorado &nbsp;&bull;&nbsp; <b>Asignatura:</b> Algoritmos Numéricos &nbsp;&bull;&nbsp; <b>Semestre:</b> 2026-2</p>
</div>
"""
display(HTML(banner_html))

def cargar_cuaderno_con_filtro(ruta_cuaderno, contenedor_vbox):
    if not os.path.exists(ruta_cuaderno):
        contenedor_vbox.children = [widgets.HTML(f"<div class='card-panel' style='color:#ef4444; font-weight:600;'>❌ Archivo no encontrado: {ruta_cuaderno}</div>")]
        return

    with open(ruta_cuaderno, 'r', encoding='utf-8') as f:
        nb = nbformat.read(f, as_version=4)
        
    lista_widgets = []
    
    for cell in nb.cells:
        source_clean = cell.source.strip()
        
        if not source_clean:
            continue
        if "import sys" in source_clean or "sys.path.append" in source_clean or "import os" in source_clean:
            continue
        if source_clean.startswith("# OMITIR") or source_clean.startswith("# HIDE"):
            continue
            
        if cell.cell_type == 'markdown':
            out_md = widgets.Output()
            with out_md:
                display(Markdown(cell.source))
            
            tarjeta_text = widgets.VBox([out_md])
            tarjeta_text.add_class("card-panel")
            lista_widgets.append(tarjeta_text)
            
        elif cell.cell_type == 'code':
            out_exec = widgets.Output()
            
            btn_run = widgets.Button(description="▶ Calcular", layout=widgets.Layout(margin='6px'))
            btn_run.add_class("btn-action")
            
            btn_clear = widgets.Button(description="🗑️ Limpiar", layout=widgets.Layout(margin='6px'))
            btn_clear.add_class("btn-reset")
            
            codigo_html = f"<div class='code-display'><b style='color:#38bdf8;'>Python Code:</b><br><br>{source_clean}</div>"
            
            def vincular_ejecucion(codigo, out_w):
                return lambda b: ejecutar_codigo_limpio(codigo, out_w)
                
            btn_run.on_click(vincular_ejecucion(cell.source, out_exec))
            btn_clear.on_click(lambda b, out_w=out_exec: out_w.clear_output())
            
            tarjeta_code = widgets.VBox([
                widgets.HTML(codigo_html),
                widgets.HBox([btn_run, btn_clear], layout=widgets.Layout(margin='0px 0px 10px 0px')),
                out_exec
            ])
            tarjeta_code.add_class("card-panel")
            lista_widgets.append(tarjeta_code)
            
    contenedor_vbox.children = lista_widgets

def ejecutar_codigo_limpio(codigo, widget_salida):
    with widget_salida:
        widget_salida.clear_output()
        display(HTML("<div class='console-output'>"))
        
        contexto_ejecucion = dict(globals())
        
        try:
            import numpy as _np
            import math as _math
            contexto_ejecucion['np'] = _np
            contexto_ejecucion['math'] = _math
            
            import Sistemas_Lineales
            import Sistemas_No_Lineales
            import Interpolacion_Lagrange
            import interpolacion_newton
            import spline_metricas
            
            for modulo in [Sistemas_Lineales, Sistemas_No_Lineales, Interpolacion_Lagrange, interpolacion_newton, spline_metricas]:
                for attr in dir(modulo):
                    if not attr.startswith('_'):
                        contexto_ejecucion[attr] = getattr(modulo, attr)
            
            exec(codigo, contexto_ejecucion)
            
        except Exception as e:
            print(f"❌ Error en los datos ingresados: {e}")
        display(HTML("</div>"))

pestana_lineales = widgets.VBox()
pestana_no_lineales = widgets.VBox()
pestana_lagrange = widgets.VBox()
pestana_newton_spline = widgets.VBox()

tabs = widgets.Tab()
tabs.children = [pestana_lineales, pestana_no_lineales, pestana_lagrange, pestana_newton_spline]
tabs.set_title(0, 'Sist. Lineales')
tabs.set_title(1, 'Sist. No Lineales')
tabs.set_title(2, 'Lagrange')
tabs.set_title(3, 'Newton / Splines')

display(tabs)

cargar_cuaderno_con_filtro("Cuaderno1_Sistemas_Lineales.ipynb", pestana_lineales)
cargar_cuaderno_con_filtro("Cuaderno2_Sistemas_No_Lineales.ipynb", pestana_no_lineales)
cargar_cuaderno_con_filtro("Cuaderno3_Interpolacion_lagrange.ipynb", pestana_lagrange)
cargar_cuaderno_con_filtro("Cuaderno4_interpolacion_newton.ipynb", pestana_newton_spline)
```
