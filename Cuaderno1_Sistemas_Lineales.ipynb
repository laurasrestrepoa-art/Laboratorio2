## ** Ejercicio 1: Bloque 07: Sistema Lineal Orden 2 (Armadura Metálica Triangular)**

### **Contexto de Ingeniería Civil**
Un ingeniero civil analiza una armadura metálica triangular $ABC$ (triángulo equilátero, base $AB = 3\text{ m}$, ángulos de $60^\circ$) destinada al soporte de una cubierta[cite: 1, 2]. Se aplica una carga vertical descendente $P = 1\text{ kN}$ en el nudo $C$[cite: 1, 2]. 

Las ecuaciones de equilibrio estático en el nudo $C$ ($\sum F_x = 0$, $\sum F_y = 0$) forman el sistema $2 \times 2$[cite: 2]:
$$\begin{cases} -0.5 F_{AC} + 0.5 F_{BC} = 0 \\ -\frac{\sqrt{3}}{2} F_{AC} - \frac{\sqrt{3}}{2} F_{BC} = 1 \end{cases}$$

En forma matricial $A\vec{x} = \vec{b}$[cite: 2]:
$$\begin{bmatrix} -0.5 & 0.5 \\ -\frac{\sqrt{3}}{2} & -\frac{\sqrt{3}}{2} \end{bmatrix} \begin{bmatrix} F_{AC} \\ F_{BC} \end{bmatrix} = \begin{bmatrix} 0 \\ 1 \end{bmatrix}$$



###  **Resumen del Desarrollo y Comprobación Manual (Laura & Dana)**

1. **Planteamiento Físico:** Se estableció el equilibrio estático ($\Sigma F_x = 0$, $\Sigma F_y = 0$) en el nudo $C$ bajo una carga vertical de referencia $P = 1\text{ kN}$.
2. **Pivoteo y Eliminación:** Se partió de la matriz aumentada $2 \times 2$ y se aplicaron operaciones de fila (normalización y escalamiento) usando $a_{11} = -0.5$ como pivote.
3. **Sustitución Regresiva:** Se obtuvo la solución exacta $F_{AC} = F_{BC} = -\frac{\sqrt{3}}{3} \approx -0.5774\text{ kN}$. El signo negativo confirmó que ambas barras trabajan a **compresión pura**.
4. **Comprobación Analítica:** Se reemplazaron las fuerzas en las ecuaciones originales de equilibrio confirmando que satisfacen las identidades $0 = 0$ y $1 = 1\text{ kN}$.
5. **Carga Máxima Admisible:** Por proporcionalidad lineal del sistema, se relacionó el esfuerzo unitario con la compresión máxima admisible ($1.5\text{ kN}$), concluyendo que $P_{\text{máx}} = \frac{1.5}{0.5774} \approx 2.60\text{ kN}$.

```python
# Planteamiento del Sistema 2x2
A_1 = np.array([
    [-0.5, 0.5],
    [-np.sqrt(3)/2.0, -np.sqrt(3)/2.0]
])
b_1 = np.array([0.0, 1.0])

# Resolución invocando la función importada de Sistemas_Lineales.py
res_1 = eliminacion_gaussiana(A_1, b_1)

if res_1['exito']:
    F_AC, F_BC = res_1['solucion']
    print(f"Solución obtenida por kN de carga P:")
    print(f"F_AC = {F_AC:.6f} kN")
    print(f"F_BC = {F_BC:.6f} kN\n")
    
    reportar_residual(res_1, unidades_b=["kN", "kN"])
    
    # Validación con NumPy
    sol_np_1 = np.linalg.solve(A_1, b_1)
    print(f"Diferencia frente a np.linalg.solve: {np.linalg.norm(res_1['solucion'] - sol_np_1):.6e}")
else:
    print(res_1['mensaje'])
```

### **Interpretación Física y Carga Máxima Admisible**
* **Estado de Esfuerzo:** Como $F_{AC} \approx -0.577350\text{ kN}$ y $F_{BC} \approx -0.577350\text{ kN}$ son negativos, ambas barras se encuentran trabajando en **compresión**.
* **Carga Máxima $P_{\text{máx}}$:**
  - Límite a tensión: $2.0\text{ kN}$.
  - Límite a compresión: $1.5\text{ kN}$
  - Dado que la compresión por kN de carga es de $0.577350\text{ kN/kN}$:
  $$P_{\text{máx}} = \frac{1.5\text{ kN}}{0.577350\text{ kN/kN}} = 2.598076\text{ kN} \approx 2.60\text{ kN}$$
  
Ambas barras gobiernan simultáneamente la capacidad del diseño por simetría.

## **Ejericio 2: Bloque 07: Sistema Lineal Orden 3 (Red de Distribución Hidráulica - Intercambio de Filas**

### **Contexto de Ingeniería Civil**
En una red de distribución de agua se determinan los caudales $Q_1, Q_2, Q_3\text{ (L/s)}$ que circulan por tres tuberías mediante balances de pérdida de carga lineal ($\Delta H = R \cdot Q$).
$$
\begin{cases}
0Q_1 + 2Q_2 + 3Q_3 = 8 \\
4Q_1 + 6Q_2 + 7Q_3 = -3 \\
2Q_1 + 6Q_2 + 6Q_3 = 5
\end{cases}
$$

Dado que el coeficiente $A_{11} = 0$, el algoritmo debe realizar un **intercambio obligatorio de filas** ($F_1 \leftrightarrow F_2$) para seleccionar $A_{21} = 4.0$ como pivote principal.

###  **Resumen del Desarrollo y Comprobación Manual (Laura & Dana)**

1. **Planteamiento de Pérdida de Carga:** Se estructuró la matriz aumentada $3 \times 3$ del balance de pérdidas de carga en los tres circuitos cerrados de la red.
2. **Intercambio Obligatorio de Filas:** Dado que el primer elemento diagonal era $A_{11} = 0$, fue necesario ejecutar un intercambio de filas ($F_1 \leftrightarrow F_2$) para seleccionar $4.0$ como elemento pivote principal.
3. **Eliminación y Sustitución Regresiva:** Se eliminaron los elementos de la columna 1 y 2 hasta obtener la matriz triangular superior, despejando secuencialmente los caudales exactos]:
   - $Q_3 = \frac{11}{4} = \mathbf{2.75\text{ L/s}}$
   - $Q_2 = -\frac{1}{8} = \mathbf{-0.125\text{ L/s}}$
   - $Q_1 = -\frac{43}{8} = \mathbf{-5.375\text{ L/s}}$
4. **Comprobación e Interpretación Física:** Se verificaron las ecuaciones por sustitución directa. Los signos negativos en $Q_1$ y $Q_2$ indican que el flujo real en esas tuberías circula en sentido contrario al supuesto en el modelo inicial.

```python
# Planteamiento del Sistema 3x3
A_2 = np.array([
    [0.0, 2.0, 3.0],
    [4.0, 6.0, 7.0],
    [2.0, 6.0, 6.0]
])
b_2 = np.array([8.0, -3.0, 5.0])

# Resolución invocando la función importada de Sistemas_Lineales.py
res_2 = eliminacion_gaussiana(A_2, b_2)

if res_2['exito']:
    Q1, Q2, Q3 = res_2['solucion']
    print(f"Solución de Caudales:")
    print(f"Q1 = {Q1:.6f} L/s")
    print(f"Q2 = {Q2:.6f} L/s")
    print(f"Q3 = {Q3:.6f} L/s\n")
    
    # Historial del pivoteo e intercambio de filas
    for paso in res_2['pasos']:
        print(f"Etapa {paso['etapa']}: Pivote = {paso['pivote_elegido']}")
        if paso['fila_intercambiada'] is not None:
            print(f"  -> Intercambio de filas ejecutado con la fila {paso['fila_intercambiada'] + 1}")
    print()
    
    reportar_residual(res_2, unidades_b=["m", "m", "m"])
    
    # Validación con NumPy
    sol_np_2 = np.linalg.solve(A_2, b_2)
    print(f"Diferencia frente a np.linalg.solve: {np.linalg.norm(res_2['solucion'] - sol_np_2):.6e}")
else:
    print(res_2['mensaje'])
```

## ** Ejecicio 3: Bloque 07: Sistema Lineal Orden 4 (Red de 4 Tanques y Nudos)**

### **Contexto de Ingeniería Civil**
Un tanque elevado $R$ ($H_R = 50\text{ m}$) y un tanque secundario $T$ ($H_T = 30\text{ m}$) alimentan una red de 4 nudos ($A, B, C, D$) con demandas $D_A = 4$, $D_B = 10$, $D_C = 15$, $D_D = 6\text{ L/s}$.
Las relaciones de conductancia $Q = C \cdot (H_i - H_j)$ conducen al sistema $4 \times 4$ en las cargas $H_A, H_B, H_C, H_D$

$$
\begin{bmatrix}
6.5 & -2.0 & -1.5 & 0.0 \\
2.0 & -5.5 & 1.0 & 2.5 \\
1.5 & 1.0 & -4.5 & 2.0 \\
0.0 & 2.5 & 2.0 & -6.5
\end{bmatrix}
\begin{bmatrix}
H_A \\ H_B \\ H_C \\ H_D
\end{bmatrix}
=
\begin{bmatrix}
146.0 \\ 10.0 \\ 15.0 \\ -54.0
\end{bmatrix}
$$

###  **Resumen del Desarrollo y Comprobación Manual (Laura & Dana)**

1. **Continuidad Nodal:** Se plantearon las 4 ecuaciones de conservación de masa ($\sum Q_{\text{entran}} = \text{Demanda}$) en los nudos $A, B, C, D$ a partir de la conductancia lineal $Q = C \cdot (H_i - H_j)$].
2. **Eliminación Gaussiana con Pivoteo:** Se seleccionó $a_{11} = 6.5$ por tener la mayor magnitud absoluta en la primera columna, escalonando la matriz mediante multiplicadores por fila].
3. **Sustitución Regresiva:** Se determinaron las cargas hidráulicas nodales:
   - $H_D \approx 29.30\text{ m}$
   - $H_C \approx 29.51\text{ m}$
   - $H_B \approx 30.97\text{ m}$
   - $H_A \approx 38.80\text{ m}$
4. **Comprobación por Balance Global:** Se calcularon los caudales tramo a tramo comprobando que el agua total aportada por los tanques $R$ y $T$ ($Q_{RA} + Q_{TD} = 33.60 + 1.40 = \mathbf{35.00\text{ L/s}}$) coincide exactamente con la demanda sumada de los nudos ($\mathbf{35\text{ L/s}}$), validando la solución matemática.

```python
# Planteamiento del Sistema 4x4
A_3 = np.array([
    [6.5, -2.0, -1.5, 0.0],
    [2.0, -5.5, 1.0, 2.5],
    [1.5, 1.0, -4.5, 2.0],
    [0.0, 2.5, 2.0, -6.5]
])
b_3 = np.array([146.0, 10.0, 15.0, -54.0])

# Resolución invocando la función importada de Sistemas_Lineales.py
res_3 = eliminacion_gaussiana(A_3, b_3)

if res_3['exito']:
    HA, HB, HC, HD = res_3['solucion']
    print("Cargas Hidráulicas Nodales:")
    print(f"H_A = {HA:.4f} m")
    print(f"H_B = {HB:.4f} m")
    print(f"H_C = {HC:.4f} m")
    print(f"H_D = {HD:.4f} m\n")
    
    reportar_residual(res_3, unidades_b=["L/s", "L/s", "L/s", "L/s"])
    
    # Validación con NumPy
    sol_np_3 = np.linalg.solve(A_3, b_3)
    print(f"Diferencia frente a np.linalg.solve: {np.linalg.norm(res_3['solucion'] - sol_np_3):.6e}\n")
    
    # Cálculo de Caudales por Tubería
    C_RA, C_AB, C_AC, C_BC, C_BD, C_CD, C_TD = 3.0, 2.0, 1.5, 1.0, 2.5, 2.0, 2.0
    H_R, H_T = 50.0, 30.0
    
    Q_RA = C_RA * (H_R - HA)
    Q_AB = C_AB * (HA - HB)
    Q_AC = C_AC * (HA - HC)
    Q_BC = C_BC * (HB - HC)
    Q_BD = C_BD * (HB - HD)
    Q_CD = C_CD * (HC - HD)
    Q_TD = C_TD * (H_T - HD)
    
    print("=" * 60)
    print("CAUDALES EN LAS TUBERÍAS (L/s):")
    print("=" * 60)
    print(f"Q_RA (R -> A) = {Q_RA:+.4f} L/s")
    print(f"Q_AB (A -> B) = {Q_AB:+.4f} L/s")
    print(f"Q_AC (A -> C) = {Q_AC:+.4f} L/s")
    print(f"Q_BC (B -> C) = {Q_BC:+.4f} L/s")
    print(f"Q_BD (B -> D) = {Q_BD:+.4f} L/s")
    print(f"Q_CD (C -> D) = {Q_CD:+.4f} L/s")
    print(f"Q_TD (T -> D) = {Q_TD:+.4f} L/s")
    
    # Balance Global de Masa
    Q_fuentes = Q_RA + Q_TD
    D_total = 4.0 + 10.0 + 15.0 + 6.0
    print("-" * 60)
    print(f"Aporte Total Fuentes (Q_RA + Q_TD): {Q_fuentes:.4f} L/s")
    print(f"Demanda Total Nudos (A+B+C+D)  : {D_total:.4f} L/s")
    print(f"Diferencia de Balance Global   : {abs(Q_fuentes - D_total):.6e} L/s")
    print("=" * 60)
else:
    print(res_3['mensaje'])
```
